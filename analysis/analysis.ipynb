```python
import pandas as pd

df = pd.read_csv("results/monza_telemetry.csv")

df.head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

plt.plot(df["time"], df["soc"])

plt.xlabel("Time (seconds)")
plt.ylabel("Battery State of Charge")

plt.title("SOC vs Time (Monza Lap)")

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.plot(df["time"], df["power"])

plt.xlabel("Time (seconds)")
plt.ylabel("MGU-K Power (W)")

plt.title("Power Deployment vs Time")

plt.show()
```

```python
print("Telemetry summary")

print("Min SOC:", df["soc"].min())
print("Max SOC:", df["soc"].max())

print("Max power deployed:", df["power"].max())
print("Average power:", df["power"].mean())
```

```python
plt.figure(figsize=(12,6))

plt.subplot(2,1,1)
plt.plot(df["time"], df["soc"])
plt.title("SOC vs Time")

plt.subplot(2,1,2)
plt.plot(df["time"], df["power"])
plt.title("Power Deployment")

plt.tight_layout()

plt.show()
```
