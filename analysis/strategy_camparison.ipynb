```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../results/experiment_results.csv")

df.head()
```

```python
for strat in df.controller.unique():
    subset = df[df.controller == strat]
    plt.plot(subset.initial_soc, subset.lap_time, marker="o", label=strat)

plt.legend()
plt.xlabel("Initial SOC")
plt.ylabel("Lap Time")
plt.title("ERS Strategy Comparison")
plt.show()
```
